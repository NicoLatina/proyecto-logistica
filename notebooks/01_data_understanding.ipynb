## Dataset Overview

El dataset contiene **3.204 registros y 23 variables**. Cada registro representa un SKU individual dentro del depósito, identificado de forma única mediante la variable `item_id`.

Las variables disponibles permiten analizar distintas dimensiones de la operación. `category` identifica la categoría del producto, mientras que `stock_level` representa el stock disponible actualmente. `reorder_point` indica el nivel de inventario a partir del cual debería iniciarse una reposición y `reorder_frequency_days` refleja la frecuencia habitual de reposición. Por su parte, `lead_time_days` representa el tiempo transcurrido entre la realización del pedido de reposición y su recepción.

En cuanto a la demanda, `daily_demand` representa la demanda diaria promedio del producto y `demand_std_dev` mide su variabilidad mediante la desviación estándar. Además, `forecasted_demand_next_7d` contiene una estimación de la demanda esperada durante los próximos siete días. `item_popularity_score` es un indicador de la popularidad relativa de cada producto.

Para analizar la operación física del depósito, el dataset incluye `storage_location_id`, que identifica la ubicación de almacenamiento del SKU, y `zone`, que indica la zona del depósito a la que pertenece. `picking_time_seconds` representa el tiempo necesario para realizar el picking de una unidad, mientras que `layout_efficiency_score` proporciona una medida de la eficiencia de la disposición de los productos dentro del almacén.

También se dispone de información económica. `handling_cost_per_unit` representa el costo asociado a la manipulación de una unidad, `unit_price` su valor unitario y `holding_cost_per_unit_day` el costo de mantener una unidad almacenada durante un día.

Respecto al nivel de servicio, `stockout_count_last_month` indica la cantidad de quiebres de stock registrados durante el último mes, `order_fulfillment_rate` representa la proporción de pedidos satisfechos correctamente y `total_orders_last_month` indica la cantidad total de órdenes registradas durante ese período.

Finalmente, `turnover_ratio` mide la rotación del inventario, `last_restock_date` registra la fecha de la última reposición y `KPI_score` funciona como un indicador general del desempeño operativo asociado a cada producto.

### Variables relevantes para los objetivos del proyecto

Para analizar el **riesgo de quiebre de stock**, las principales variables serán `stock_level`, `reorder_point`, `lead_time_days`, `daily_demand`, `demand_std_dev`, `forecasted_demand_next_7d` y `stockout_count_last_month`. Estas variables permitirán comparar las existencias actuales con la demanda esperada, los tiempos de reposición y el comportamiento histórico de cada SKU.

Para evaluar el **desempeño de las distintas zonas del depósito**, serán especialmente relevantes `zone`, `storage_location_id`, `layout_efficiency_score`, `picking_time_seconds` e `item_popularity_score`. Esto permitirá estudiar si los productos de mayor movimiento están ubicados de manera eficiente y si existen zonas con tiempos de picking superiores al resto.

Para analizar el **costo y valor del inventario**, se utilizarán principalmente `stock_level`, `unit_price`, `holding_cost_per_unit_day` y `handling_cost_per_unit`.

Para detectar posibles casos de **sobrestock**, será necesario relacionar `stock_level` con `daily_demand`, `forecasted_demand_next_7d`, `reorder_point`, `lead_time_days`, `demand_std_dev` y `turnover_ratio`. Un nivel elevado de stock no necesariamente implica sobrestock si el producto posee una demanda alta o tiempos de reposición prolongados.

Por último, para investigar **tiempos elevados de picking**, las variables principales serán `picking_time_seconds`, `storage_location_id`, `zone`, `item_popularity_score`, `layout_efficiency_score` e `item_id`, con el objetivo de detectar productos o ubicaciones que puedan estar afectando la eficiencia operativa.
